```python
import nltk
from nltk.stem import PorterStemmer, RegexpStemmer, SnowballStemmer, LancasterStemmer, WordNetLemmatizer
```

```python
stemming = PorterStemmer()
```

```python
words = ["running", "runner", "ran", "runs", "easily", "fairly", "happily", "happiness", "happily", "happier", "happiest","enumerate","enumeration","enumerator","eat","eating","eaten","eats","swim","swimming","swimmer","swims","swum","swam","swum","swimming","swimmer","swims","swum","swam","swum","swimming","swimmer","swims","swum","swam","swum"]
```

```python
for word in words:
    print("Original ---> Stemmed:")
    print(word, "--->", stemming.stem(word))
```

```python
regexp_stemming = RegexpStemmer('ing$|s$|e$|able$', min=4)
for word in words:
    print("Original ---> Regexp Stemmed:")
    print(word, "--->", regexp_stemming.stem(word))
```

```python
snowball_stemming = SnowballStemmer("english")
for word in words:
    print("Original ---> Snowball Stemmed:")
    print(word, "--->", snowball_stemming.stem(word))
```

```python
for word in words:
    print("Original ---> Stemming,Regexp Stemmed,Snowball Stemmed:")
    print(word, "--->", stemming.stem(word), regexp_stemming.stem(word), snowball_stemming.stem(word))
```
