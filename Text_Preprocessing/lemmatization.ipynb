```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
words = ["running", "runner", "ran", "runs", "easily", "fairly", "happily", "happiness", "happily", "happier", "happiest","enumerate","enumeration","enumerator","eat","eating","eaten","eats","swim","swimming","swimmer","swims","swum","swam","swum","swimming","swimmer","swims","swum","swam","swum","swimming","swimmer","swims","swum","swam","swum"]
```

```python
''' 
POS
Noun - n
Verb - v
Adjective - a
Adverb - r
'''
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize(words[0], pos='v'),lemmatizer.lemmatize(words[1], pos='n'),lemmatizer.lemmatize("happily", pos='r')

```

```python
for word in words:
    print("Original :", word)
    print("Lemmatized (Verb):", lemmatizer.lemmatize(word, pos='v'))
    print("Lemmatized (Noun):", lemmatizer.lemmatize(word, pos='n'))
    print("Lemmatized (Adjective):", lemmatizer.lemmatize(word, pos='a'))
    print("Lemmatized (Adverb):", lemmatizer.lemmatize(word, pos='r'))
```
