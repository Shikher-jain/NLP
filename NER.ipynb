```python
import nltk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
# NER - Named Entity Recognition
sentence = "The Effiel Tower is located in Paris, France. It was built in 1889 and is one of the most famous landmarks in the world. The tower was designed by the French engineer Gustave Eiffel and stands at a height of 324 meters. It attracts millions of visitors each year who come to admire its unique architecture and breathtaking views of the city. The Effiel Tower is not only a symbol of Paris but also a testament to human ingenuity and engineering prowess. It has become an iconic representation of French culture and is often featured in movies, literature, and art. Whether you visit during the day or at night when it is beautifully illuminated, the Effiel Tower never fails to leave a lasting impression on those who see it. It is truly a marvel of modern engineering and a must-see destination for anyone traveling to Paris."
```

```python
words = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)  
```

```python
ne_chunks = nltk.ne_chunk(pos_tags)
print(ne_chunks)
```
